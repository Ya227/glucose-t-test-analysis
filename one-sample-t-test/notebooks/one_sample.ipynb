# Одновыборочный критерий Стьюдента
Анализ уровня глюкозы у диабетиков и проверка гипотезы о среднем значении.
Автор: Карасёва Яна

```python
import pandas as pd
import numpy as np
import statistics
from scipy import stats
import matplotlib.pyplot as plt
```

## Формирование выборки
Выбираем только тех пациентов, у которых диабет = 1.

```python
url = "https://raw.githubusercontent.com/ya227/glucose-t-test-analysis/main/one-sample-t-test/data/diabetes.csv"
df = pd.read_csv(url)

glucose = df.iloc[:300, 1].tolist()
diabet = df.iloc[:300, 8].tolist()
data = []
for i in range(len(glucose)):
    if(diabet[i]==1):
        data.append(glucose[i])

data.sort()
```

## Описательная статистика
Рассчитываем основные метрики выборки.

```python
n = len(data)
print('Объем выборки: ', n)
sr = statistics.mean(data)
print("Среднее арифметическое: ", sr)
max_g = data[-1]
print("Максимум: ", max_g)
min_g = data[0]
print("Минимум: ", min_g)
dis = statistics.variance(data)
print("дисперсия: ", dis)
med = statistics.median(data)
print("медиана: ", med)
st = statistics.stdev(data)
print("стандартное отклонение: ", st)
moda = statistics.mode(data)
print("мода: ", moda)
```

## Одновыборочный t‑тест
Проверяем гипотезу о равенстве среднего значению 130.

```python
alpha = 0.05
beta = 1-alpha
df = n - 1
T = ((sr-130)/st)*(np.sqrt(df))
critical_value = stats.t.ppf(1 - alpha / 2, df)
p_value = 2 * stats.t.sf(abs(T), df)
print('Статистика одновыборочного критерия Стьюдента:', T)
print(f"Критическое значение (C): {critical_value}")
print(f"p-значение: {p_value}")
```

```python
x=np.linspace(-4,4,500)
y=stats.t.pdf(x, df)

plt.figure(figsize=(10,5))
plt.plot(x, y, 'b-', label=f'T (df={df})')

plt.fill_between(x, y, where=(x <= -critical_value), color='red', alpha=0.3, label='Критическая область')
plt.fill_between(x, y, where=(x >= critical_value), color='red', alpha=0.3)

plt.axvline(T, color='g', linestyle='--', label=f'Выборочная статистика (T={T:.2f})')
plt.title('Двусторонний T критерий Стьюдента')
plt.xlabel('Значение t-статистики')
plt.ylabel('Плотность вероятности')
plt.legend()
plt.grid(True)
plt.show()
```

# Выводы
- t‑статистика превышает критическое значение;
- p‑value меньше уровня значимости;
- гипотеза H0 отвергается;
- средний уровень глюкозы у диабетиков статистически отличается от 130.
